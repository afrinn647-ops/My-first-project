```python
import numpy as np
import matplotlib.pyplot as plt
from statistics import multimode
```

```python
# dataset
marks=[50,52,54,56,58,
       60,62,64,66,68,
       70,72,74,75,76,
       78,80,82,84,86,
       88,90,92,94,96,
       54,56,58,60,62,
       64,66,68,70,72,
       74,75,76,78,80,
       82,84,86,88,90,
       92,94,96,98,100
       ]
#bins
bins= np.arange(50,105,5)
```

```python
# descriptive statistics
mean_marks= np.mean(marks)
median_marks=np.median(marks)
mode_marks=multimode(marks)

print(f"Mean={mean_marks}")
print(f"Median={median_marks}")
print(f"Mode={mode_marks}")
```

```python
# for histogram
plt.figure(figsize=(10,8))
plt.hist(marks, bins=bins, edgecolor='green',alpha=0.8)
plt.title("Histogram of math exam score")
plt.xlabel("Score range")
plt.ylabel("Frequency")
plt.xticks(np.arange(50,105,5))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# KDE distribution plot
# KDE plot (Estimated probability density function)
plt.figure(figsize=(10,6))
sns.kdeplot(
    marks,
    color='blue',
    linewidth=3,
    fill=True,
    alpha=0.5,
    label="Estimated PDF (KDE)"
)

plt.title("distribution of math exam score")
plt.xlabel("scores")
plt.ylabel('density')
plt.xticks(np.arange(50,105,5))
plt.legend()
plt.show()
```

```python
import numpy as np
# dataset
scores=np.array([-200, 50, 52,54,56,58,60,62,64,66,68,70,72,74,75,76,78,80,82,84,86,88,90,94,100,500])
mean=np.mean(scores)
std=np.std(scores)
print(mean,std)
```

```python
Z_score=(scores-mean)/std
np.set_printoptions(suppress=True)
print(Z_score)
```

```python
low_outliers=scores[Z_score <-3]
high_outlier=scores[Z_score > 3]
print(low_outliers)
print(high_outlier)
```

```python
# combine both sides
outliers_z=np.concatenate([low_outliers,high_outlier])
print(outliers_z)
if len(outliers_z) >0:
  print("Z_score mathod outliers:", outliers_z)
else:
  print("no outliers found")
```
