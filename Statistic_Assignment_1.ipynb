```python
import numpy as np
from statistics import multimode
import matplotlib.pyplot as plt
```

```python
# Showing mean, median, mode, varience and Std from both datasets

# Calling both the datasets
secA_scores=np.array([15, 40, 23, 19, 12, 12, 11, 30, 19, 22, 10, 45, 28, 12, 12, 12, 14, 17, 16, 13, 19, 12, 13, 15, 16, 25,
12, 17, 19, 90])
secB_scores=np.array([35, 38, 90, 50, 85, 83, 87, 62, 64, 90, 90, 80, 70, 72, 74, 75, 76, 78, 80, 82, 84, 85, 86, 88, 90, 92,
94, 96, 98])
# mean
meanA=np.mean(secA_scores)
meanB=np.mean(secB_scores)
print(f"mean of secA dataset: {meanA}")
print(f"mean of secB dataset: {meanB}")

# median

medianA=np.median(secA_scores)
medianB=np.median(secB_scores)
print(f"median of secA dataset: {medianA}")
print(f"median of secB dataset: {medianB}")

# mode
modeA=multimode(secA_scores)
modeB=multimode(secB_scores)
print(f"mode of secA dataset: {modeA}")
print(f"mode of secB dataset: {modeB}")

# varience
varienceA=np.var(secA_scores)
varienceB=np.var(secB_scores)
print(f"varience of secA dataset: {varienceA}")
print(f"varience of secB dataset: {varienceB}")

# STd
stdA=np.std(secA_scores)
stdB=np.std(secB_scores)
print(f"STD of secA dataset: {stdA}")
print(f"STD of secB dataset: {stdB}")
```

Explain why mean > mode in a right-skewed dataset

In a right-skewed distribution, most values are small and clustered on the left.
A few large values stretch the tail to the right, pulling the mean toward them.
The mean is affected by these extreme values, so it shifts higher.
The mode stays at the peak where most data points occur, unaffected by the tail.
As a result, in right skewness, the mean is greater than the mode.
for example, if we see at secA_score dataset, it has a mean around 20.67 while mode is only 12.

```python
import numpy as np
secA_scores=np.array([15, 40, 23, 19, 12, 12, 11, 30, 19, 22, 10, 45, 28, 12, 12, 12, 14, 17, 16, 13, 19, 12, 13, 15, 16, 25,
12, 17, 19, 90])
secB_scores=np.array([35, 38, 90, 50, 85, 83, 87, 62, 64, 90, 90, 80, 70, 72, 74, 75, 76, 78, 80, 82, 84, 85, 86, 88, 90, 92,
94, 96, 98])
```

```python
# Looking for outliers in those datasets
# for dataset 1

Z_scoreA=(secA_scores-meanA)/stdA
np.set_printoptions(suppress=True)
print(Z_scoreA)

low_outliersA = secA_scores[Z_scoreA < -3]
high_outliersA = secA_scores[Z_scoreA > 3]

print(low_outliersA)
print(high_outliersA)

outliers_zA=np.concatenate([low_outliersA,high_outliersA])
print(outliers_zA)
if len(outliers_zA) >0:
  print("Z_score mathod outliers:", outliers_zA)
else:
  print("no outliers found")
```

```python
# for dataset 2
Z_scoreB=(secB_scores-meanB)/stdB
np.set_printoptions(suppress=True)
print(Z_scoreB)

low_outliersB = secB_scores[Z_scoreB < -3]
high_outliersB = secB_scores[Z_scoreB > 3]

print(low_outliersB)
print(high_outliersB)

outliers_zB=np.concatenate([low_outliersB,high_outliersB])
print(outliers_zB)
if len(outliers_zB) >0:
  print("Z_score mathod outliers:", outliers_zB)
else:
  print("no outliers found")
```

```python
# Loading tips dataset from seaborn
import seaborn as snb
tips=snb.load_dataset("tips")
print(tips)
```

```python
# female customers who tipped $3 or less.

import scipy.stats as stats

female_tips = tips[tips['sex'] == 'Female']['tip']
percentile_rank = stats.percentileofscore(female_tips, 3)
print(percentile_rank)
```

```python
# female customers who tipped more than $3

female_tips = tips[tips['sex'] == 'Female']['tip']
percentage_above_3 = (female_tips > 3).mean() * 100
print(percentage_above_3)
```

```python
# male customers who smoke and tipped more than $5
male_smokers_high_tip = tips[(tips['sex'] == 'Male') & (tips['smoker'] == 'Yes') & (tips['tip'] > 5)]
print(male_smokers_high_tip)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew
```

```python
# Extract total_bill column
total_bill = tips['total_bill']

# Plot histogram with KDE
plt.figure(figsize=(10,6))
sns.histplot(total_bill, kde=True, bins=20, color='green')
plt.title('Histogram of Total Bill with KDE')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()

# Calculate skewness
total_bill_skew = skew(total_bill)
print("Skewness of total_bill:", total_bill_skew)

# Interpretation
if total_bill_skew > 0:
    print("The distribution is right-skewed (positively skewed).")
elif total_bill_skew < 0:
    print("The distribution is left-skewed (negatively skewed).")
else:
    print("The distribution is approximately symmetric (normal-like).")
```

```python
# 5-number summary
five_number_summary = tips['total_bill'].describe()[['min', '25%', '50%', '75%', 'max']]
print("5-number summary:\n", five_number_summary)

# Quantiles
quantiles = tips['total_bill'].quantile([0.25, 0.5, 0.75])
print("\nQuantiles:\n", quantiles)

# Interquartile Range (IQR)
iqr = quantiles[0.75] - quantiles[0.25]
print("\nInterquartile Range (IQR):", iqr)
```
